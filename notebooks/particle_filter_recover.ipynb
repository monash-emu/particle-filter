```python
import numpy as np
import pandas as pd

from emu_filter.state_model import predict_states
from emu_filter.importance import truncnorm_importance
from emu_filter.outputs import get_counts_from_particles, plot_particle_results, get_links_from_pedigree, plot_links
```

```python
# Inputs
params = {
    "contact_rate": 1.0,
    "recovery_rate": 0.2,
}
suscept_init = 990
inf_init = 10
n_particles = 20
n_obs = 30
target_sd = 25.0

init_state = np.array([suscept_init, inf_init, 0])
n_states = init_state.size
```

```python
# Simulate from parameters
sim_states = np.zeros([n_obs, n_states], dtype=int)
sim_states[0, :] = init_state

state = init_state
for o in range(n_obs):
    state_reshaped = state.reshape(1, -1)
    state = predict_states(state_reshaped, **params)
    sim_states[o, :] = state
observations = sim_states[:, 1]
```

```python
# Initialise particles
particles = np.zeros([n_particles, n_obs + 1, n_states], dtype=int)
particles[:, 0, :] = init_state
prop_particles = np.copy(particles)
pedigree = np.zeros([n_particles, n_obs], dtype=int)

# Main loop
for o, obs in enumerate(observations):

    # Prediction
    particle_states = particles[:, o, :]
    proposed_particles = predict_states(particle_states, **params)
    prop_particles[:, o + 1] = proposed_particles

    # Importance
    weights = truncnorm_importance(proposed_particles[:, 1], obs, target_sd)
    norm_weights = weights / sum(weights)

    # Resampling
    indices = np.random.choice(n_particles, size=n_particles, p=norm_weights)
    pedigree[:, o] = indices
    resamp_particles = proposed_particles[indices, :]
    
    # Update
    particles[:, o + 1, :] = resamp_particles
```

```python
plot_particle_results(prop_particles, particles, observations)
```

```python
links = get_links_from_pedigree(particles, pedigree, observations)
plot_links(particles, links, observations)
```
